# Phase 2: Predicting negative customer reviews

**Our research question:** Can we predict whether a delivered order will receive a negative review (1–2 stars)?

**Our stakeholder:** We focus on the customer experience and seller quality teams, who could use predicted risk to prioritise follow-up after delivery and before a review is submitted.

**Our target:** We define `is_negative_review = 1` for a 1–2 star review and `0` for a 3–5 star review.

**Our selected features:** We use 21 order-level predictors, grouped by the information they represent:

| Feature group [column names] | Why we include them |
|---|---|
| Delivery performance [`delivery_days`, `days_early`, `late_delivery_flag`] | We investigate whether delivery duration and arrival relative to the promised date are associated with dissatisfaction. Negative `days_early` means late; zero means on time; positive means early. |
| Order value and payment [`order_item_value`, `order_freight_value`, `payment_total`, `payment_count`, `max_installments`, `freight_share`] | We examine whether order cost, shipping cost and payment characteristics help explain review risk. |
| Order and seller complexity [`item_count`, `seller_count`, `n_seller_states`] | We test whether orders involving more items or sellers have different review outcomes. |
| Product characteristics [`primary_product_category`, `product_weight_g_mean`, `product_volume_cm3_mean`, `product_photos_qty_mean`] | We examine whether the product mix and physical product characteristics contribute predictive information. |
| Geography [`customer_state`, `seller_state`, `same_state`] | We account for broad customer and seller location differences. |
| Purchase timing [`purchase_month`, `purchase_weekday`] | We allow for seasonal and weekday variation in order experiences. |

These are **candidate predictors**, not proven causes of negative reviews. In our modelling, we evaluate whether they help predict the target rather than assuming every feature is important.

**Our approach:** We compare Logistic Regression, Decision Tree and Random Forest, beginning with simple baselines. We split orders chronologically into an earlier training period and a later test period. Within the training period, we use time-ordered validation folds to tune hyperparameters, select our model **and choose the decision threshold** for customer contact. Only then do we evaluate the frozen model and threshold, once, on the test period.

**Our prediction point:** We predict immediately after delivery, before a review has been submitted. We therefore exclude orders that were already reviewed before delivery. We do not use review scores, text or timestamps as predictors, because they would reveal information about the outcome.

## 1. We prepare our data and define the target

We build **one row per delivered order** with a known review. We define `is_negative_review = 1` for 1–2 stars and `0` for 3–5 stars.

We use our existing `build_delivered_order_classification_dataset()` function to assemble validated order-level features. Our inputs describe delivery performance, order value, product details and geography.

**Why we filter:** We remove orders whose `review_answer_timestamp` occurred before `order_delivered_customer_date`. Our proposed model predicts at delivery, so it cannot intervene before a review that already exists. We use these two timestamps only for this historical eligibility check, **not** as model features.

In the code below, we print the population before and after filtering so we can explain exactly which orders we model.

```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.base import clone
from sklearn.metrics import (
    average_precision_score, precision_score, recall_score, f1_score,
    roc_auc_score, roc_curve, precision_recall_curve, confusion_matrix,
    make_scorer, ConfusionMatrixDisplay,
)
from sklearn.inspection import permutation_importance
from sklearn.model_selection import TimeSeriesSplit, cross_validate, GridSearchCV

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from src.classification_data import (
    TARGET_COLUMN, DELIVERED_MODEL_FEATURES, build_delivered_order_classification_dataset,
    split_delivered_order_features_target,
)
from src.classification_model import build_delivered_order_logistic_pipeline
from src.classification_tree_model import (
    build_delivered_order_decision_tree_pipeline,
    build_delivered_order_random_forest_pipeline,
)

# We fix the random seed so our results can be reproduced.
SEED = 42

# We load our Phase 1 processed tables.
orders = pd.read_csv(ROOT / "data/processed/order_level.csv")
items = pd.read_csv(ROOT / "data/processed/item_level.csv")

# We reuse the team's validated order-level feature construction.
df = build_delivered_order_classification_dataset(orders, items)

# We align original order timestamps with our modelling rows.
lookup = orders.set_index("order_id").loc[df["order_id"]]
df["purchase_time"] = pd.to_datetime(lookup["order_purchase_timestamp"], format="mixed").to_numpy()

# We identify reviews submitted before delivery.
reviewed_at = pd.to_datetime(lookup["review_answer_timestamp"], format="mixed").to_numpy()
delivered_at = pd.to_datetime(lookup["order_delivered_customer_date"], format="mixed").to_numpy()

# We remove customers who had already reviewed before our prediction point.
already_reviewed = reviewed_at < delivered_at

# We report the original population and our exclusions.
print(f"Delivered orders with review: {len(df):,}")
print(f"Excluded (review before delivery): {already_reviewed.sum():,}")

# We retain eligible orders in chronological order.
df = df.loc[~already_reviewed].sort_values("purchase_time", kind="stable").reset_index(drop=True)
print(f"Eligible orders: {len(df):,}; negative review rate: {df[TARGET_COLUMN].mean():.1%}")

display(df[TARGET_COLUMN].value_counts().rename(index={0:"3–5 stars",1:"1–2 stars"}).to_frame("orders"))
```

## 2. We check leakage and split chronologically

We make predictions **after delivery, before review submission**. That lets us use delivery outcomes such as `days_early` and `late_delivery_flag`, but not information from the review itself. We use `split_delivered_order_features_target()` to keep the feature contract explicit.

**Our feature availability check:** We use `order_delivered_customer_date` to establish the delivery event and derive delivery performance; we do not feed the raw timestamp to the model. This is valid **only because our prediction point is after delivery**. If we were predicting at checkout, `delivery_days`, `days_early` and `late_delivery_flag` would leak future outcomes. We do not use `review_score`, `review_creation_date` or `review_answer_timestamp` as predictors. We derive the target from `review_score` and use `review_answer_timestamp` only to exclude orders that had already reviewed before delivery.

**Our temporal limitation:** We split by purchase time, while our prediction point occurs at delivery. This prevents a random mixture of earlier and later purchases, but does not guarantee that every training order's label was observable before later orders were scored. We acknowledge this limitation instead of claiming that the split eliminates all temporal leakage.

We sort orders by `purchase_time` and reserve the newest 20% for our final test. We train and tune on the earlier 80%. This is a chronological proxy, not a complete simulation of every production event time.

**How we use each period:** this is the train–validation–test methodology the project brief asks for.

| Role | Which orders | What we use them for |
|---|---|---|
| **Training folds** | The earlier part of each of five forward-chaining splits inside the first 80% | Fitting each model |
| **Validation folds** | The later part of each split, always after its training fold | Comparing models, tuning hyperparameters and **choosing the decision threshold** (sections 3–5) |
| **Test period** | The newest 20% of orders | One final evaluation of the frozen model and threshold (section 6 onwards) |

We keep imputation, scaling and categorical encoding inside the scikit-learn pipelines. This matters because preprocessing statistics must be learnt from each training fold, not from future data.

```python
# We select the approved predictors and keep the review label separate.
X, y = split_delivered_order_features_target(df)
assert X.columns.tolist() == DELIVERED_MODEL_FEATURES
assert not any("review" in col for col in X.columns)

# We reserve the newest 20% as our untouched test period.
cut = int(len(df) * 0.80)
X_train, X_test = X.iloc[:cut], X.iloc[cut:]
y_train, y_test = y.iloc[:cut], y.iloc[cut:]

# We quantify imbalance in the eligible modelling population.
negative_rate = y.mean()
majority_accuracy = max(negative_rate, 1 - negative_rate)
print(f"Eligible negative-review rate: {negative_rate:.1%}")
print(f"Majority-class accuracy reference: {majority_accuracy:.1%}")
print("This reference predicts only the most common class and misses the minority class.")

# We show sample sizes and negative-review rates for each period.
display(pd.DataFrame([
    {"population": "Train", "orders": len(y_train), "negative_rate": y_train.mean(),
     "first_purchase": df["purchase_time"].iloc[0].date(),
     "last_purchase": df["purchase_time"].iloc[cut-1].date()},
    {"population": "Test", "orders": len(y_test), "negative_rate": y_test.mean(),
     "first_purchase": df["purchase_time"].iloc[cut].date(),
     "last_purchase": df["purchase_time"].iloc[-1].date()},
]).round(3))
```

## 3. We compare simple baselines using cross-validation

We begin with plain Logistic Regression as an **interpretable linear baseline** and a single Decision Tree as the **simplest tree-based baseline**. We then compare a default Random Forest to test whether combining many trees improves on one tree. These represent **two model families**, consistent with our IT5006 model-family budget. We use the same feature set and evaluation folds for a fair comparison.

We use `TimeSeriesSplit(n_splits=5)` **within the training period**. The validation folds always follow their corresponding training folds in time.

### Why we use these evaluation metrics

Our target is imbalanced: relatively few delivered orders receive a 1–2 star review. We report the rate for the **eligible modelling population**, not just the unfiltered delivered orders. The following code displays that rate and a majority-class accuracy reference so we can see why accuracy alone would be misleading. We therefore do not use overall accuracy as our main measure, because a model could appear accurate simply by predicting non-negative reviews for most orders.

| Metric | What we learn from it | How we use it |
|---|---|---|
| **PR AUC** (`average_precision`) | Across possible decision thresholds, how well do we rank and retrieve rare negative-review cases while maintaining precision? | **Our primary CV model-selection metric.** |
| **Precision** (`precision`) | Of the orders we flag as at risk, what proportion actually receive a negative review? | We assess unnecessary customer follow-ups. |
| **Recall** (`recall`) | Of the orders that receive a negative review, what proportion do we flag? | We assess how many dissatisfied customers we miss. |
| **F1** (`f1`) | How well do we balance precision and recall at a single threshold? | We report one complementary threshold-based summary. |
| **ROC AUC** (`roc_auc`) | How well do we rank a randomly chosen negative-review order above a non-negative one? | We check overall discrimination, while recognising it can look favourable with imbalanced data. |

We choose **PR AUC** because our main challenge is finding a relatively rare negative class and we have not fixed an operational contact threshold. In scikit-learn, we calculate this using **average precision** (`average_precision`), which summarises the precision–recall curve; it is not exactly the same as trapezoidal PR curve area.

We report precision, recall and F1 at the **illustrative 0.50 classification threshold**. Changing that threshold would change these three values, but not PR AUC or ROC AUC. In section 5, we replace 0.50 with a threshold chosen on the validation folds.

Our aim is to test whether added model complexity improves meaningful predictive performance, not to maximise the number of metrics or algorithms.

**Checking for overfitting:** we also record each model's score on the training folds it learned from, so we can compare training, validation and test performance in section 6.1.

**Undefined precision:** When a model predicts no negative reviews in a validation fold, precision has no positive predictions to evaluate. We explicitly use `make_scorer(precision_score, zero_division=0)` to report zero rather than repeatedly printing warnings. This changes neither fitted predictions nor our PR AUC model-selection criterion.

```python
# We perform time-ordered validation only on earlier training orders.
cv = TimeSeriesSplit(n_splits=5)

# We evaluate ranking and threshold-based performance.
metrics = {"precision":make_scorer(precision_score, zero_division=0),"recall":"recall","f1":"f1","roc_auc":"roc_auc","pr_auc":"average_precision"}

# We start with simple representatives from the two model families.
baseline = {
    "Logistic (plain)": build_delivered_order_logistic_pipeline(random_state=SEED).set_params(classifier__C=np.inf),
    "Decision Tree (plain)": build_delivered_order_decision_tree_pipeline(random_state=SEED),
    "Random Forest (default)": build_delivered_order_random_forest_pipeline(random_state=SEED, n_estimators=200),
}

# We compare the simple models using the same time-ordered folds.
# We also keep training-fold scores (to check overfitting) and the fold-to-fold spread (for model selection).
rows = []
for name, pipeline in baseline.items():
    result = cross_validate(pipeline, X_train, y_train, cv=cv, scoring=metrics, n_jobs=1, return_train_score=True)
    rows.append({"model":name, **{metric:result[f"test_{metric}"].mean() for metric in metrics},
                 "pr_auc_std":result["test_pr_auc"].std(), "train_pr_auc":result["train_pr_auc"].mean()})
baseline_cv = pd.DataFrame(rows).set_index("model")

display(baseline_cv[list(metrics)].round(3))
```

## 4. We tune our two main models

We tune only the two model families we plan to compare: Logistic Regression and Random Forest.

We use small, explainable parameter grids:

| Model | Parameter | Values tested | Why we vary it |
|---|---|---|---|
| Logistic Regression | `classifier__C` | `0.01, 0.1, 1, 10` | We test stronger versus weaker regularisation to control coefficient complexity. |
| Logistic Regression | `classifier__class_weight` | `None, balanced` | We test whether giving more weight to negative reviews improves performance. |
| Random Forest | `classifier__max_depth` | `12, None` | We compare constrained trees with unrestricted depth. |
| Random Forest | `classifier__min_samples_leaf` | `10, 25` | We limit how few training orders can form a leaf, reducing reliance on very small groups. |
| Random Forest | `classifier__class_weight` | `None, balanced_subsample` | We test whether weighting minority cases within each tree's bootstrap sample helps. |

We deliberately test **8 Logistic Regression combinations** and **8 Random Forest combinations**, each across five chronological CV folds. These are limited illustrative grids, not proven optimal ranges. We can explain every parameter without adding more algorithms.

**How we handle imbalance:** We include both `class_weight=None` and class-weighted alternatives in our training-only search. Weighting gives more influence to the minority class during fitting. We do not assume weighting is better: we compare the cross-validation scores and report which setting was selected for each model. We do not add oversampling or SMOTE without evidence that it would improve our approach.

We use `GridSearchCV` with `refit="pr_auc"`, so the best settings are chosen from **training cross-validation PR AUC**. We do not use the held-out test set to choose parameters or select our model. We print the best settings and a concise comparison table.

### How we choose between models

The project brief asks that any more complex model **demonstrate improvement** over a simpler one. We apply this with the **one-standard-error rule** (Hastie, Tibshirani & Friedman, 2009, *The Elements of Statistical Learning*, section 7.10). We find the model with the highest CV PR AUC and calculate the standard error of its score (the standard deviation across the five folds divided by √5). We then choose the **simplest** model whose CV PR AUC lies within one standard error of the best. A smaller difference is within the noise of our cross-validation estimate, so it does not justify a more complex model.

We rank complexity by family: Logistic Regression is the simplest, then a single Decision Tree, then Random Forest. If two candidates are equally simple, we take the one with the higher CV PR AUC. This rule uses training cross-validation results only.

```python
# We tune only on training folds, using PR AUC to select settings.
logistic_grid = GridSearchCV(
    build_delivered_order_logistic_pipeline(random_state=SEED),
    {"classifier__C":[0.01,0.1,1.0,10.0],
     "classifier__class_weight":[None,"balanced"]},
    cv=cv, scoring=metrics, refit="pr_auc", n_jobs=1, return_train_score=True,
)

# We search a compact set of Random Forest settings.
forest_grid = GridSearchCV(
    build_delivered_order_random_forest_pipeline(random_state=SEED,n_estimators=200),
    {"classifier__max_depth":[12,None],
     "classifier__min_samples_leaf":[10,25],
     "classifier__class_weight":[None,"balanced_subsample"]},
    cv=cv, scoring=metrics, refit="pr_auc", n_jobs=1, return_train_score=True,
)

# We fit both searches using training orders only.
logistic_grid.fit(X_train,y_train)
forest_grid.fit(X_train,y_train)

# We summarise the selected settings and CV scores.
tuned = {"Logistic (tuned)":logistic_grid, "Random Forest (tuned)":forest_grid}
print("Selected parameters (training CV only):")
for name, search in tuned.items():
    print(name, search.best_params_)
    selected_weight = search.best_params_["classifier__class_weight"]
    weighted_options = pd.DataFrame(search.cv_results_)[
        ["param_classifier__class_weight", "mean_test_pr_auc"]
    ]
    # We compare the best PR AUC available with and without class weighting.
    weighted_options["weight_setting"] = weighted_options[
        "param_classifier__class_weight"
    ].map(lambda value: "None" if pd.isna(value) else str(value))
    comparison = (
        weighted_options.groupby("weight_setting")["mean_test_pr_auc"]
        .max()
        .rename("best_cv_pr_auc")
    )
    print(f"Selected class weight: {selected_weight!r}")
    display(comparison.round(3).to_frame())

tuned_cv = pd.DataFrame([
    {"model":name, **{metric:search.cv_results_[f"mean_test_{metric}"][search.best_index_] for metric in metrics},
     "pr_auc_std":search.cv_results_["std_test_pr_auc"][search.best_index_],
     "train_pr_auc":search.cv_results_["mean_train_pr_auc"][search.best_index_]}
    for name,search in tuned.items()
]).set_index("model")
cv_results = pd.concat([baseline_cv,tuned_cv])

display(cv_results.sort_values("pr_auc",ascending=False)[list(metrics)].round(3))

# We lock our model choice before examining the test results, using the one-standard-error rule.
complexity = {"Logistic (plain)":0, "Logistic (tuned)":0, "Decision Tree (plain)":1,
              "Random Forest (default)":2, "Random Forest (tuned)":2}
best_name = cv_results["pr_auc"].idxmax()
standard_error = cv_results.loc[best_name,"pr_auc_std"] / np.sqrt(cv.get_n_splits())
within_one_se = cv_results.loc[cv_results["pr_auc"] >= cv_results.loc[best_name,"pr_auc"] - standard_error]
chosen_name = (within_one_se.assign(complexity=within_one_se.index.map(complexity))
               .sort_values(["complexity","pr_auc"], ascending=[True,False]).index[0])
print(f"Highest CV PR AUC: {best_name} ({cv_results.loc[best_name,'pr_auc']:.4f}; one standard error = {standard_error:.4f})")
print("Models within one standard error of the best:", ", ".join(within_one_se.index))
print("Selected model (simplest within one standard error):", chosen_name)
```

## 5. We select an intervention threshold on the validation folds, before testing

A model outputs a **probability** of a negative review. To decide whom to contact, we also need a **decision threshold**: we flag an order when its probability is at or above that cutoff. The default 0.50 is not necessarily appropriate for an imbalanced target. In the validation folds of section 4, tuned Logistic Regression recalled only about **2%** of negative reviews at 0.50, and tuned Random Forest recalled almost none.

We treat the threshold as part of the model's decision rule, so we choose it the same way we chose hyperparameters: **on validation data only, before we open the test set**. The test period is then used once, in section 6, to evaluate the complete frozen decision rule (model, hyperparameters and threshold). Choosing the cutoff after looking at test results would make the test score optimistic.

**How we create validation predictions:** We reuse the five forward-chaining folds from sections 3 and 4. For each fold, we refit the complete tuned pipeline on the earlier orders and predict only that fold's later validation orders. Combining the five folds gives one validation probability per order, for every training order except the earliest block, which is never a validation fold.

**Our threshold policy:** Our illustrative operational policy is to **maximise recall subject to at least 20% precision** on the combined validation predictions. In plain terms: catch as many unhappy customers as possible, while at least one in five contacted customers really is unhappy. This is a stated scenario, not a stakeholder-approved cost threshold. If no threshold meets the precision floor, we fall back to the threshold with the highest validation F1 and flag the policy as infeasible. We choose a threshold separately for each tuned finalist, so we compare both at their own operating points.

"Optimal" here means optimal **for this policy**. A different contact budget or cost assumption would give a different threshold.

We test cutoffs from 0.01 to 0.99 in steps of 0.01 and plot how precision, recall and F1 change as the cutoff moves.

```python
# We score any cutoff with the same counts: flagged orders, detected and missed negative reviews, false alarms.
def threshold_metrics(y_true, p, cutoff):
    pred = (p >= cutoff).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
    return {"precision": precision_score(y_true, pred, zero_division=0),
            "recall": recall_score(y_true, pred, zero_division=0),
            "f1": f1_score(y_true, pred, zero_division=0),
            "flagged_share": float(pred.mean()), "false_alarms": int(fp),
            "missed_negative_reviews": int(fn), "detected_negative_reviews": int(tp)}

# We reuse the five time-ordered folds; each validation fold always follows its training fold.
validation_folds = list(cv.split(X_train))
fold_id = np.concatenate([np.full(len(valid_idx), fold)
                          for fold, (_, valid_idx) in enumerate(validation_folds, start=1)])
y_val = np.concatenate([y_train.iloc[valid_idx].to_numpy() for _, valid_idx in validation_folds])
assert len(np.unique(y_val)) == 2, "Validation orders must include both classes."

threshold_finalists = ["Logistic (tuned)", "Random Forest (tuned)"]
threshold_candidates = np.unique(np.r_[np.linspace(0.01, 0.99, 99), 0.50])
precision_floor = 0.20  # Illustrative operating constraint, not a stakeholder requirement.
validation_probs, threshold_rows, threshold_choices = {}, [], {}

for name in threshold_finalists:
    # We refit the tuned pipeline on each fold's earlier orders and score only its later validation orders.
    validation_probs[name] = np.concatenate([
        clone(tuned[name].best_estimator_).fit(X_train.iloc[fit_idx], y_train.iloc[fit_idx])
        .predict_proba(X_train.iloc[valid_idx])[:, 1]
        for fit_idx, valid_idx in validation_folds
    ])

    # We measure the effect of every candidate cutoff on the combined validation orders.
    options = pd.DataFrame([{"model": name, "threshold": float(cutoff),
                             **threshold_metrics(y_val, validation_probs[name], cutoff)}
                            for cutoff in threshold_candidates])
    threshold_rows.append(options)

    # We apply our policy: the highest recall with at least 20% precision, otherwise the highest F1.
    eligible = options.loc[options["precision"] >= precision_floor]
    feasible = not eligible.empty
    ranking = (eligible.sort_values(["recall", "precision", "threshold"], ascending=False) if feasible
               else options.sort_values(["f1", "precision", "threshold"], ascending=False))
    threshold_choices[name] = {"threshold": float(ranking.iloc[0]["threshold"]), "policy_feasible": feasible}

threshold_analysis = pd.concat(threshold_rows, ignore_index=True)
print(f"Validation orders used to choose the threshold: {len(y_val):,} of {len(y_train):,} training orders")
display(pd.DataFrame(threshold_choices).T)

# We plot how precision, recall and F1 respond as the cutoff moves.
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, name in zip(axes, threshold_finalists):
    rows = threshold_analysis.loc[threshold_analysis["model"] == name]
    ax.plot(rows["threshold"], rows["recall"], label="Recall")
    ax.plot(rows["threshold"], rows["precision"], label="Precision")
    ax.plot(rows["threshold"], rows["f1"], label="F1")
    ax.axhline(precision_floor, color="grey", linestyle="--", linewidth=1, label="20% precision floor")
    ax.axvline(threshold_choices[name]["threshold"], color="black", linestyle=":", label="Selected threshold")
    ax.set(title=name + " (validation folds)", xlabel="Decision threshold", ylabel="Score", ylim=(0, 1))
    ax.legend()
    ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

### 5.1 How our policy arrives at 0.12

We choose the threshold in four steps, all on the validation folds:

1. **We score the validation orders.** Every validation order gets a predicted probability from a model that never saw it (the out-of-fold predictions above).
2. **We try every cutoff.** For each cutoff from 0.01 to 0.99, in steps of 0.01, we flag the orders whose probability is at or above the cutoff. We compare the flags with the actual reviews and count detected negative reviews, false alarms and missed negative reviews. From these counts we calculate precision, recall and F1.
3. **We drop the cutoffs that break the precision floor.** We keep only cutoffs where at least 20% of the flagged customers actually left a negative review.
4. **We pick the remaining cutoff with the highest recall.** Raising the cutoff can only remove orders from the flagged list, so recall can only stay the same or fall as the cutoff rises. Step 4 therefore picks, in effect, **the lowest cutoff that still meets the 20% floor**. If two cutoffs catch exactly the same number of negative reviews, we take the one with higher precision.

The table below shows part of this search: the round cutoffs 0.10, 0.20, 0.30 and 0.40, the cutoffs either side of the selected value, and the default 0.50. **These are validation results.** We do not build this table on the test period, because choosing a cutoff by comparing test results would make the test score optimistic.

```python
# We show part of the validation search: round cutoffs, neighbours of the selected cutoff, and 0.50.
shown_cutoffs = [0.10, 0.11, 0.12, 0.13, 0.20, 0.30, 0.40, 0.50]
selected_cutoff = {name: round(choice["threshold"], 2) for name, choice in threshold_choices.items()}
threshold_table = threshold_analysis.assign(threshold=threshold_analysis["threshold"].round(2))
threshold_table = threshold_table.loc[threshold_table["threshold"].isin(shown_cutoffs)].copy()
threshold_table["flagged_orders"] = (threshold_table["detected_negative_reviews"]
                                     + threshold_table["false_alarms"])
threshold_table["meets_20pct_floor"] = threshold_table["precision"] >= precision_floor
threshold_table["note"] = np.select(
    [threshold_table["threshold"] == threshold_table["model"].map(selected_cutoff),
     threshold_table["threshold"] == 0.50],
    ["Selected", "Default"], default="",
)
display(threshold_table.set_index(["model", "threshold"])[
    ["precision", "recall", "f1", "flagged_share", "flagged_orders", "detected_negative_reviews",
     "false_alarms", "missed_negative_reviews", "meets_20pct_floor", "note"]
].round(3))

# For comparison only: the cutoff a different policy (highest F1) would have chosen.
for name in threshold_finalists:
    rows = threshold_analysis.loc[threshold_analysis["model"] == name]
    best_f1 = rows.loc[rows["f1"].idxmax()]
    print(f"{name}: highest validation F1 is {best_f1['f1']:.3f}, at threshold {best_f1['threshold']:.2f}")
```

**Reading the table:**

* **Each row answers one question: "if we contacted every customer at or above this cutoff, what would happen?"** For example, at 0.10, tuned Logistic Regression would contact **17,360** validation customers (**28.6%**) and reach **2,977** of the **5,909** who left a negative review (recall **50.4%**). But **14,383** of those contacts would be false alarms (precision **17.1%**).
* **Raising the cutoff shortens the contact list: precision rises and recall falls.** For Logistic Regression, moving from 0.10 to 0.20, 0.30 and 0.40 raises precision from **17.1%** to **36.3%**, **45.4%** and **51.2%**, while recall falls from **50.4%** to **15.0%**, **8.4%** and **4.9%**. At 0.50, the list shrinks to **257** customers (**0.4%**) and recall to **2.3%**.
* **Why 0.12, and not 0.10 or 0.11:** at 0.10 and 0.11, Logistic Regression's precision (**17.1%** and **19.3%**) is below the 20% floor, so step 3 drops them. At 0.12, precision first clears the floor (**21.6%**). Every higher cutoff also clears it but catches fewer negative reviews (for example **1,828** at 0.13 against **2,129** at 0.12), so step 4 picks 0.12.
* **The floor is what stops us from going lower.** Lowering the cutoff from 0.12 to 0.10 would reach **848** more unhappy customers, but would add **6,674** contacts to customers who were satisfied: about eight extra false alarms for each extra unhappy customer reached.
* **The same steps give 0.12 for the forest, with less margin.** Its precision is **18.3%** at 0.11 and **20.1%** at 0.12, only just above the floor. This matches the per-fold warning below.
* **A different policy could give a different threshold.** For comparison, the cutoff with the highest validation F1 is also **0.12** for Logistic Regression (F1 **0.270**) but **0.15** for the forest (F1 **0.285**). A fixed contact budget, such as "call the riskiest 5% of customers", would give another cutoff again. So 0.12 is the best cutoff **for our stated policy**, not a universal optimum.

### 5.2 We check the selected threshold in each validation period

The table above combines all five validation folds. We now check each fold separately. The folds run forward in time, so a precision that holds in the latest folds gives us more confidence that the threshold will carry forward to the test period. These are still validation results: we have not opened the test set.

```python
# We check whether the selected cutoff keeps its precision in each validation period, especially the latest.
fold_rows = []
for fold, (_, valid_idx) in enumerate(validation_folds, start=1):
    in_fold = fold_id == fold
    for name in threshold_finalists:
        fold_rows.append({
            "fold": fold, "model": name,
            "first_purchase": df["purchase_time"].iloc[valid_idx[0]].date(),
            "last_purchase": df["purchase_time"].iloc[valid_idx[-1]].date(),
            "negative_rate": y_val[in_fold].mean(),
            **threshold_metrics(y_val[in_fold], validation_probs[name][in_fold],
                                threshold_choices[name]["threshold"]),
        })
fold_check = pd.DataFrame(fold_rows)
display(fold_check[["fold", "model", "first_purchase", "last_purchase", "negative_rate",
                    "precision", "recall", "flagged_share"]].round(3))

print("Thresholds frozen before opening the test set:",
      {name: choice["threshold"] for name, choice in threshold_choices.items()})
```

**What we found in each validation period:**

* **The per-fold check gives an early warning for the forest.** At 0.12, Logistic Regression kept precision between **21.6%** and **23.8%** in folds 2–5 and fell below 20% only in fold 1 (**16.4%**), the earliest fold, whose model learnt from the fewest orders. The forest fell below 20% in fold 1 (**15.4%**) and in fold 5 (**19.0%**), and sat at the floor in fold 2 (**20.0%**). Fold 5 is the latest validation period, closest to the test period.
* **The negative-review rate moved between folds** (**8.6%** to **10.9%**). Precision at a fixed cutoff depends on that rate, which is one reason a threshold can drift over time.
* **We freeze 0.12 for both models** before opening the test set.

## 6. We evaluate the frozen models and thresholds once on later orders

We now open the test period for the first time. Nothing is fitted or chosen here: we apply the model selected in section 4 and the thresholds frozen in section 5.

We report two kinds of results:

* **Ranking quality for all five models:** PR AUC and ROC AUC, which do not depend on any threshold. We also show precision, recall and F1 at 0.50, so these rows can be compared with the cross-validation tables in sections 3 and 4.
* **Contact decisions for the two tuned finalists** at their **validation-selected thresholds**, with the default 0.50 as a reference: precision, recall, F1, the share of orders flagged, false alarms and missed negative reviews. We also check whether the 20% precision floor still holds on later orders.

**ROC AUC and PR AUC do not change when we change only the decision threshold.** The threshold changes which customers we contact, not how well the model ranks them.

```python
# We train each baseline on earlier orders and keep the test set for evaluation.
fitted = {name:clone(model).fit(X_train,y_train) for name,model in baseline.items()}
fitted.update({name:search.best_estimator_ for name,search in tuned.items()})

# We calculate test probabilities and ranking metrics.
probs, test_rows = {}, []
for name,model in fitted.items():
    p = model.predict_proba(X_test)[:,1]
    # We use 0.50 here only to compare with the cross-validation tables in sections 3 and 4.
    pred = (p>=0.5).astype(int)
    probs[name] = p
    test_rows.append({
        "model":name, "precision":precision_score(y_test,pred,zero_division=0),
        "recall":recall_score(y_test,pred,zero_division=0),
        "f1":f1_score(y_test,pred,zero_division=0),
        "roc_auc":roc_auc_score(y_test,p), "pr_auc":average_precision_score(y_test,p)
    })
test_results = pd.DataFrame(test_rows).set_index("model")

# We compare results with the prevalence of negative reviews.
display(test_results.round(3))
print(f"Test prevalence, the random ranking PR reference: {y_test.mean():.3f}")
print("Model selected before opening the test set:",chosen_name)
```

```python
# We apply each frozen threshold once to the test probabilities, with 0.50 as a reference.
threshold_test_comparison = pd.DataFrame([
    {"model": name, "policy": policy, "threshold": cutoff, **threshold_metrics(y_test, probs[name], cutoff)}
    for name in threshold_finalists
    for policy, cutoff in [("Default 0.50", 0.50), ("Validation selected", threshold_choices[name]["threshold"])]
])
threshold_test_comparison["meets_precision_floor"] = threshold_test_comparison["precision"] >= precision_floor
display(threshold_test_comparison.round(3))
print(f"Negative review prevalence in the test period: {y_test.mean():.4f}")

# We compare validation and test results at the selected thresholds to see how well the choice carried forward.
val_selected = pd.DataFrame.from_dict({
    name: threshold_metrics(y_val, validation_probs[name], threshold_choices[name]["threshold"])
    for name in threshold_finalists
}, orient="index").rename_axis("model")
test_selected = threshold_test_comparison.loc[threshold_test_comparison["policy"] == "Validation selected"].set_index("model")
display(pd.DataFrame({
    "threshold": test_selected["threshold"],
    "validation_precision": val_selected["precision"], "test_precision": test_selected["precision"],
    "validation_recall": val_selected["recall"], "test_recall": test_selected["recall"],
    "validation_flagged_share": val_selected["flagged_share"], "test_flagged_share": test_selected["flagged_share"],
}).round(3))

# We plot ranking performance for the tuned finalists and mark each frozen operating point.
fig, axes = plt.subplots(1,2,figsize=(12,4))
for name in tuned:
    fpr,tpr,_ = roc_curve(y_test,probs[name])
    precision,recall,_ = precision_recall_curve(y_test,probs[name])
    line, = axes[0].plot(fpr,tpr,label=name)
    axes[1].plot(recall,precision,label=name,color=line.get_color())
    point = test_selected.loc[name]
    point_label = f"{name} at threshold {point['threshold']:.2f}"
    axes[0].scatter(point["false_alarms"] / (y_test == 0).sum(), point["recall"],
                    s=64, color=line.get_color(), edgecolor="white", zorder=3, label=point_label)
    axes[1].scatter(point["recall"], point["precision"],
                    s=64, color=line.get_color(), edgecolor="white", zorder=3, label=point_label)

# We add random-ranking references, the precision floor and readable plot labels.
axes[0].plot([0,1],[0,1],linestyle=":",color="grey",label="Random")
axes[1].axhline(y_test.mean(),linestyle=":",color="grey",label="Prevalence")
axes[1].axhline(precision_floor,linestyle="--",linewidth=1,color="grey",label="20% precision floor")
axes[0].set(title="ROC (test period)",xlabel="False positive rate",ylabel="True positive rate")
axes[1].set(title="Precision–recall (test period)",xlabel="Recall",ylabel="Precision")
for ax in axes:
    ax.legend(fontsize=8)
    ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

**What we found:**

* **Ranking quality:** The two tuned finalists ranked later orders almost identically: **PR AUC 0.187 for Logistic Regression versus 0.188 for Random Forest**, and **ROC AUC 0.649 versus 0.645**. Both are well above the test prevalence reference of **0.085**. The PR AUC difference of `0.001` gives no compelling reason to prefer the more complex forest.
* **Our selected model meets the policy on later orders.** At its frozen threshold of 0.12, tuned Logistic Regression flagged **7.6%** of test orders (**1,385**) and detected **343** of the **1,543** negative reviews: **recall 22.2%, precision 24.8%**, with **1,042** false alarms. Precision stayed above the 20% floor. At 0.50, it would have detected only **35** (recall **2.3%**).
* **The forest does not meet the policy on later orders.** At 0.12, tuned Random Forest detected more negative reviews (**525**, recall **34.0%**), but its precision fell to **18.6%**, below the floor, with **2,297** false alarms. The latest validation fold had already shown the forest's precision slipping below 20%.
* **The threshold carried forward, but not exactly.** At the same 0.12 cutoff, both models flagged fewer orders in the test period than in validation: **7.6%** versus **16.2%** for Logistic Regression and **15.5%** versus **21.2%** for Random Forest. Recall was therefore lower than validation suggested (**22.2%** versus **36.0%** for Logistic Regression). For Logistic Regression, the shorter list was more precise (**24.8%** versus **21.6%**). A fixed probability cutoff assumes that scores keep the same meaning over time. Possible reasons they did not here: the test period has fewer negative reviews (**8.5%** versus **9.7%** in training), and the final model learnt from more orders than each validation model.
* These results describe how our frozen choices performed. We do not go back and change the threshold or the model because of them.

### 6.1 We compare training, validation and test performance

We compare each model's PR AUC on the training folds it learned from (**train**), on the later cross-validation folds it did not see (**validation**), and on the held-out later orders (**test**). A model that scores much higher on training data than on validation data is memorising its training orders rather than learning patterns that carry over to new orders. Because our folds follow time, each validation fold covers a different period, so we focus on the size of the train–validation gap rather than on small differences.

```python
# We compare training, validation and test PR AUC to check for overfitting.
fit_check = pd.DataFrame({
    "train_pr_auc": cv_results["train_pr_auc"],
    "validation_pr_auc": cv_results["pr_auc"],
    "test_pr_auc": test_results["pr_auc"],
})
fit_check["train_minus_validation"] = fit_check["train_pr_auc"] - fit_check["validation_pr_auc"]
display(fit_check.loc[test_results.index].round(3))
```

**What we found:**

* **Neither Logistic Regression model shows signs of overfitting.** Their training PR AUC (0.201 plain, 0.193 tuned) is close to, or slightly below, their validation PR AUC (0.205 and 0.218). A training score slightly below the validation score is not a problem here: each fold's training data covers different, earlier periods, and PR AUC depends on how common negative reviews are in each period.
* **The single Decision Tree and the default Random Forest memorise their training orders.** Both score a perfect 1.000 on training folds, but only 0.105 and 0.203 on validation.
* **Tuning reduces the forest's overfitting but does not remove it.** The tuned Random Forest scores 0.334 on training folds against 0.229 on validation, a gap of 0.105.
* **Every model scores lower on the later test period than in validation**, for example 0.218 → 0.187 for tuned Logistic Regression. The test period also has fewer negative reviews (8.5%, against 9.7% in training). On test, the two tuned models end up almost identical (0.187 vs 0.188).
* This supports our selection: tuned Logistic Regression generalises as well as the tuned Random Forest, with far less overfitting.

## 7. We interpret what the model relies on

We use `permutation_importance()` on the selected model and report the 10 inputs with the biggest average decrease in `average_precision` when shuffled.

We interpret these values as **how strongly the fitted model depends on an input for prediction**, not proof that an input causes customers to leave low reviews.

We keep this analysis descriptive. We do not revisit hyperparameters based on test-set feature importance.

### What our feature analysis suggests

In our run, the selected tuned Logistic Regression relied most on `item_count` and `seller_count`: shuffling them decreased test PR AUC by `0.0392` and `0.0373`. `delivery_days` followed (`0.0251`). Every other input had a decrease of `0.004` or less, including `late_delivery_flag` (`0.0015`) and `days_early` (`0.0006`).

This suggests that **order complexity and delivery duration are the main predictive signals** for customers who had not yet reviewed at delivery. Lateness carries much less signal here than our Phase 1 analysis suggested. One possible reason is that many late orders were reviewed before delivery and are excluded from this population. We cannot conclude that additional items or sellers *cause* negative reviews. These figures come from three shuffles per input, so very small values should not be over-interpreted.

```python
# We shuffle each input to measure its effect on held-out PR AUC.
importance = permutation_importance(
    fitted[chosen_name], X_test, y_test, scoring="average_precision",
    n_repeats=3, random_state=SEED, n_jobs=1
)

# We rank the inputs by their average impact on PR AUC.
feature_importance = pd.Series(
    importance.importances_mean,index=X_test.columns,name="PR AUC decrease"
).sort_values(ascending=False)
display(feature_importance.head(10).round(4).to_frame())
```

## 8. We inspect both models' errors at their selected thresholds

We compare the tuned Logistic Regression and Random Forest confusion matrices at the thresholds frozen in section 5, not at 0.50. We then describe the detected negative reviews, missed negative reviews and false alarms by lateness, delivery time and order size. We include order size because `item_count` and `seller_count` were the inputs our selected model relied on most (section 7).

This is a descriptive look at the test period. We do not change the thresholds or the model based on it.

```python
# We compare both tuned models at their validation-selected thresholds.
test_flags = {name: (probs[name] >= threshold_choices[name]["threshold"]).astype(int)
              for name in threshold_finalists}
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

for ax, name in zip(axes, threshold_finalists):
    matrix = confusion_matrix(y_test, test_flags[name], labels=[0, 1])
    ConfusionMatrixDisplay(
        confusion_matrix=matrix,
        display_labels=["3–5 stars", "1–2 stars"],
    ).plot(ax=ax, cmap="Blues", colorbar=False, values_format=",d")
    ax.set_title(f"{name} (threshold {threshold_choices[name]['threshold']:.2f})")
    ax.set_xlabel("Predicted review")
    ax.set_ylabel("Actual review")

fig.suptitle("Test-period confusion matrices at the validation-selected thresholds")
plt.tight_layout()
plt.show()

# We compare error patterns for both models.
error_summaries = []
for name in threshold_finalists:
    errors = X_test.copy()
    errors["actual"] = y_test.to_numpy()
    errors["predicted"] = test_flags[name]
    errors["result"] = np.select(
        [(errors["actual"] == 1) & (errors["predicted"] == 1),
         (errors["actual"] == 1) & (errors["predicted"] == 0),
         (errors["actual"] == 0) & (errors["predicted"] == 1)],
        ["Detected negative", "Missed negative", "False alarm"], default="Correctly not flagged",
    )
    summary = errors.groupby("result").agg(
        orders=("actual", "size"),
        late_share=("late_delivery_flag", "mean"),
        median_delivery_days=("delivery_days", "median"),
        multi_item_share=("item_count", lambda count: (count > 1).mean()),
    ).reset_index()
    summary.insert(0, "model", name)
    error_summaries.append(summary)

display(pd.concat(error_summaries, ignore_index=True).round(3))
```

**What we found:**

* **Both models mostly flag multi-item orders.** For tuned Logistic Regression, **82.5%** of detected negative reviews and **67.0%** of false alarms had more than one item, compared with **4.1%** of the orders it correctly left alone. This matches section 7, where `item_count` and `seller_count` mattered most.
* **The forest flags almost every multi-item order.** Among orders the forest did not flag, almost none had more than one item (**0.0%** of those correctly not flagged and **0.1%** of missed negative reviews). This is why it flags more orders than Logistic Regression and produces more false alarms.
* **Flagged orders are more often late and slower.** For Logistic Regression, **8.2%** of detected negative reviews and of false alarms were late, against **1.1%** of the orders it correctly left alone. False alarms had the longest median delivery time (**11** days).
* **The missed negative reviews look like ordinary orders.** Of the **1,200** negative reviews Logistic Regression missed, only **9.3%** had more than one item and only **1.6%** were late (median delivery time **8** days). Our inputs give the model little to separate these customers from satisfied ones. The validation curves in section 5 show that a lower threshold would catch more of them only by pushing precision below our 20% floor.

## 9. We conclude and explain the limitations

**Our model selection:** Using training cross-validation only, tuned Random Forest had the highest PR AUC (**0.229**), but tuned Logistic Regression (**0.218**) was within one standard error (**0.020**) of it. Following our one-standard-error rule, we **select tuned Logistic Regression**, the simpler model.

**Our threshold, chosen before testing:** We treated the decision threshold as part of the model and chose it on the validation folds only, before opening the test set. Under our illustrative policy (maximise recall with at least 20% precision), both tuned models received a threshold of **0.12**. On validation, tuned Logistic Regression at 0.12 detected **36.0%** of negative reviews at **21.6%** precision, against **2.3%** recall at 0.50. Lower cutoffs (0.10 and 0.11) broke the 20% precision floor; higher cutoffs met it but caught fewer negative reviews (recall **15.0%** at 0.20 and **4.9%** at 0.40). The per-fold check showed the forest's precision at 0.12 slipping below 20% in the latest validation fold (**19.0%**), while Logistic Regression stayed above 20% in folds 2–5.

**Our held-out test findings:** We evaluated the frozen models and thresholds once. On ranking, the tuned models were almost identical (PR AUC **0.187** versus **0.188**; ROC AUC **0.649** versus **0.645**; prevalence reference **0.085**), so the more complex forest shows no meaningful advantage. At 0.12, tuned Logistic Regression flagged **7.6%** of test orders and detected **343** negative reviews (recall **22.2%**, precision **24.8%**) with **1,042** false alarms, meeting the policy. Tuned Random Forest reached recall **34.0%**, but its precision (**18.6%**) fell below the 20% floor, as its latest validation fold had warned.

**The threshold carried forward, but not exactly:** At the same cutoff, both models flagged fewer test orders than validation orders (Logistic Regression **7.6%** versus **16.2%**), so test recall was lower than validation suggested. A fixed probability cutoff depends on scores keeping the same meaning over time, so a deployed threshold would need monitoring and periodic re-checking on recent orders.

**Overfitting:** Section 6.1 shows that tuned Logistic Regression does not overfit (training PR AUC **0.193** versus validation **0.218**). The tuned Random Forest still does (**0.334** versus **0.229**), and the default forest and the single tree memorise their training data (training PR AUC **1.000**).

**What tuning achieved:** Our observed test PR AUC increased from **0.180 to 0.187** for Logistic Regression and from **0.178 to 0.188** for Random Forest. These are descriptive held-out comparisons, **not** evidence that tuning will always improve future performance.

**How our stakeholder could use the result:** The customer experience team could score newly delivered orders that have not already been reviewed, and contact customers whose predicted risk is 0.12 or higher. In the test period, that would have meant contacting **1,385** customers, about one in four of whom left a negative review, against roughly one in twelve across all test orders. The list is dominated by orders with more than one item. Most missed negative reviews were single-item orders that arrived on time, which our inputs struggle to separate from satisfied customers. Predicted risk does not prove that an intervention would prevent a negative review.

**Our class imbalance sensitivity check:** A separate chronological validation experiment compared natural Logistic Regression training with balanced class weights and 50/50 training-only undersampling. Neither balancing approach materially improved PR AUC; see the [classification workflow documentation](../docs/phase2-classification-workflow.md) and [recorded CSV](../reports/phase2/three_approaches.csv). These results do not replace the main cross-validation comparison.

**Our limitations:** We use historical Olist data; some causes of dissatisfaction are not recorded; we exclude the 4,653 orders reviewed before delivery, so the model does not cover those customers; and our purchase-time split is not a complete reproduction of real-world scoring conditions. Our measured differences between the tuned models are small, and permutation importance uses only three shuffles per input. The 20% precision floor is an illustrative policy, not a stakeholder requirement. We used the same validation folds to tune hyperparameters and to choose the threshold, so validation results at 0.12 may be slightly optimistic; the test period was used for neither, so the test evaluation is not affected. The earliest **12,156** training orders never appear in a validation fold, so they do not inform the threshold.

**Our next step:** Before Phase 3, agree a stakeholder-defined contact budget or cost trade-off for the selected Logistic Regression, re-run the validation threshold selection under that policy, and monitor precision and the share of flagged orders on recent orders before relying on a fixed cutoff for customer outreach.
